# 07. Retention Action Engine & ROI Optimization

This notebook covers:
1. Mapping top SHAP churn risk drivers to specific retention offers (discount, tech support, fiber upgrade, e-check switch incentive).
2. Financial ROI calculation:
   $$\text{Expected ROI} = \text{CLV} \times P(\text{Churn}) \times \text{Uplift} - \text{Offer Cost}$$
3. Automated recommendation of optimal action, priority score, and budget allocation.

```python
import json
import os
import pandas as pd

pb_path = '../models/retention_playbook.json'
if os.path.exists(pb_path):
    with open(pb_path, 'r') as f:
        playbook = json.load(f)
    print(f"Loaded {len(playbook)} Retention Playbook Rules:")
    for feature, action in list(playbook.items())[:10]:
        print(f"- Feature: {feature:<30} -> Offer: {action['offer_name']} (Cost: ${action['cost']})")
else:
    print("Playbook file not found.")
```
