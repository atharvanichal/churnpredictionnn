# 03. Churn Classifier Training — Logistic Regression, LightGBM, Optuna & Calibration

This notebook documents:
1. Baseline Logistic Regression classifier.
2. LightGBM Gradient Boosting classifier with Optuna hyperparameter optimization (50 trials).
3. Probability calibration via `CalibratedClassifierCV` (isotonic regression).
4. Model evaluation: ROC-AUC, PR-AUC, F1 score, Brier score, and Expected Calibration Error.

```python
import json
import os
import pandas as pd
import joblib

metrics_path = '../models/training_metrics.json'
if os.path.exists(metrics_path):
    with open(metrics_path, 'r') as f:
        metrics = json.load(f)
    print("--- Model Training Performance Metrics ---")
    for model_name, m in metrics.items():
        print(f"\n[{model_name.upper()}]")
        for k, v in m.items():
            if isinstance(v, float):
                print(f"  {k}: {v:.4f}")
            else:
                print(f"  {k}: {v}")
else:
    print("Training metrics file not found. Ensure run_training.py completes.")
```
