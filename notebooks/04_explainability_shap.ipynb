# 04. Model Explainability — SHAP (SHapley Additive exPlanations)

This notebook covers:
1. SHAP TreeExplainer calculation for global and local churn risk factors.
2. Global feature importance visualizations (Beeswarm and Bar plots).
3. Local inference breakdown: converting top SHAP impact scores into human-readable business reason codes for actionable customer retention.

```python
import json
import os
import pandas as pd

fi_path = '../models/global_feature_importance.json'
if os.path.exists(fi_path):
    with open(fi_path, 'r') as f:
        fi = json.load(f)
    df_fi = pd.DataFrame(list(fi.items()), columns=['Feature', 'Mean |SHAP|']).sort_values('Mean |SHAP|', ascending=False)
    print("Top 15 Most Impactful Features on Churn Risk:")
    display(df_fi.head(15))
else:
    print("Feature importance file not found.")
```
